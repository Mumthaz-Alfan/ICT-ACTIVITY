```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, f1_score
from sklearn.model_selection import RandomizedSearchCV
import lightgbm as lgb

```

```python
train_df = pd.read_csv('/content/train_LZdllcl.csv')
test_df = pd.read_csv('/content/test_2umaH9m.csv')
print(train_df.shape)
print(test_df.shape)
```

```python
display(train_df.describe(include='all'))
display(test_df.describe(include='all'))
```

```python
# Check missing values separately
print("\n--- Missing Values in Train ---")
print(train_df.isnull().sum()[train_df.isnull().sum() > 0])

print("\n--- Missing Values in Test ---")
print(test_df.isnull().sum()[test_df.isnull().sum() > 0])
```

```python
print("\n--- Target Variable Distribution (is_promoted) ---")
print(train_df['is_promoted'].value_counts(normalize=True))
```

```python
# 1. Separate Features (X) and Target (y)
X = train_df.drop(columns=['employee_id', 'is_promoted'])
y = train_df['is_promoted']

# Keep Test IDs for submission file and drop from features
test_ids = test_df['employee_id']
X_test = test_df.drop(columns=['employee_id'])

# 2. Define Feature Types based on EDA
num_cols = ['no_of_trainings', 'age', 'previous_year_rating',
            'length_of_service', 'KPIs_met >80%', 'awards_won?', 'avg_training_score']
cat_cols = ['department', 'region', 'education', 'gender', 'recruitment_channel']

# 3. Pipelines (SimpleImputer fills missing NaNs dynamically so 0 rows are dropped)
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])

# 4. Train / Validation Split on Training Set (80% Train, 20% Validation)
X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 5. Fit Preprocessor ONLY on Training Data & Transform sets separately
X_train_prep = preprocessor.fit_transform(X_train)
X_val_prep = preprocessor.transform(X_val)
X_test_prep = preprocessor.transform(X_test)

# Verify no rows were lost
print("Preprocessing complete!")
print(f"Train features shape:      {X_train_prep.shape}")
print(f"Validation features shape: {X_val_prep.shape}")
print(f"Test features shape:       {X_test_prep.shape}")
```

```python
# 1. Initialize Baseline Model
# class_weight='balanced' handles the 91:9 imbalanced target
rf_baseline = RandomForestClassifier(random_state=42, class_weight='balanced')

# 2. Train on Preprocessed Training Data
rf_baseline.fit(X_train_prep, y_train)

# 3. Predict on Validation Data
y_val_pred = rf_baseline.predict(X_val_prep)

# 4. Show Evaluation Metrics
print("--- Baseline Model Performance ---")
print(f"Validation F1-Score: {f1_score(y_val, y_val_pred):.4f}\n")
print(classification_report(y_val, y_val_pred))
```

```python


# 1. Define hyperparameter options to try
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# 2. Search for the best settings using 3-Fold Cross-Validation
random_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(random_state=42, class_weight='balanced'),
    param_distributions=param_grid,
    n_iter=10,
    scoring='f1',
    cv=3,
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train_prep, y_train)

# 3. Get the Best Model
best_model = random_search.best_estimator_
print("Best Hyperparameters:", random_search.best_params_)

# 4. Predict on the Test Set (test_2umaH9m.csv)
test_preds = best_model.predict(X_test_prep)

# 5. Save the Final Submission File
submission = pd.DataFrame({
    'employee_id': test_ids,
    'is_promoted': test_preds
})

submission.to_csv('submission.csv', index=False)
print("\nPredictions saved to 'submission.csv'!")
print("Final Submission Shape:", submission.shape)
```

```python


# 1. Calculate class weights for imbalance
scale_pos_weight = (len(y_train) - sum(y_train)) / sum(y_train)

# 2. Train LightGBM Classifier
lgb_model = lgb.LGBMClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    scale_pos_weight=scale_pos_weight,
    random_state=42
)

lgb_model.fit(X_train_prep, y_train)

# 3. Evaluate on Validation Set
y_val_lgb_pred = lgb_model.predict(X_val_prep)
print("--- LightGBM Model Performance ---")
print(f"Validation F1-Score: {f1_score(y_val, y_val_lgb_pred):.4f}\n")
print(classification_report(y_val, y_val_lgb_pred))

# 4. Save Improved Test Predictions
test_lgb_preds = lgb_model.predict(X_test_prep)

submission_improved = pd.DataFrame({
    'employee_id': test_ids,
    'is_promoted': test_lgb_preds
})

submission_improved.to_csv('submission.csv', index=False)
print("Updated 'submission.csv' with LightGBM predictions!")
```

```python


# 1. Train LightGBM without heavy scale_pos_weight
lgb_opt = lgb.LGBMClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    num_leaves=31,
    random_state=42
)
lgb_opt.fit(X_train_prep, y_train)

# 2. Get probabilities & find optimal decision threshold
val_probs = lgb_opt.predict_proba(X_val_prep)[:, 1]
best_thresh, best_f1 = 0.5, 0

for thresh in np.arange(0.25, 0.45, 0.02):
    score = f1_score(y_val, (val_probs >= thresh).astype(int))
    if score > best_f1:
        best_f1, best_thresh = score, thresh

print(f"Optimal Threshold: {best_thresh:.2f}")
print(f"Improved Validation F1-Score: {best_f1:.4f}")

# 3. Save predictions using optimal threshold
test_probs = lgb_opt.predict_proba(X_test_prep)[:, 1]
sample_sub = pd.read_csv('/content/sample_submission_M0L0uXE.csv')
sample_sub['is_promoted'] = (test_probs >= best_thresh).astype(int)
sample_sub.to_csv('final_submission_av.csv', index=False)
print("Saved optimized predictions to 'final_submission_av.csv'!")
```
