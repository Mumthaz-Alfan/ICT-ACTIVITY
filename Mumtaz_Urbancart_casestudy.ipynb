**MODULE5**

```python
# 1. Get product price by product ID
def get_product_price(product_id, products_list):
  for p in products_list:
    if p["Product_ID"] == product_id:
      return p["Price"]
  return None


# 2. Get customer name by customer ID
def get_customer_name(customer_id, customers_list):
  for c in customers_list:
    if c["Customer_ID"] == customer_id:
      return c["Name"]
  return "Unknown"


# 3. Calculate order value
def calculate_order_value(price, quantity):
  return price * quantity


# 4. Classify order value using default arguments
def classify_order(value, high_limit=3000, low_limit=1000):
  if value > high_limit:
    return "High"
  elif value >= low_limit:
    return "Medium"
  else:
    return "Low"


# 5. Generate a formatted invoice for a single order
def generate_invoice(order, products_list, customers_list):
  cust_name = get_customer_name(order["Customer_ID"], customers_list)
  price = get_product_price(order["Product_ID"], products_list)
  val = calculate_order_value(price, order["Quantity"])

  invoice = f"--- INVOICE FOR {order['Order_ID']} ---\n"
  invoice += f"Customer: {cust_name}\n"
  invoice += f"Product ID: {order['Product_ID']}\n"
  invoice += f"Quantity: {order['Quantity']}\n"
  invoice += f"Total Value: ₹{val}\n"
  invoice += f"Status: {order['Status']}"
  return invoice


# 6. Calculate total revenue (excluding cancelled orders)
def calculate_total_revenue(orders_list, products_list):
  total = 0
  for order in orders_list:
    if order["Status"] == "Cancelled":
      continue
    price = get_product_price(order["Product_ID"], products_list)
    total += calculate_order_value(price, order["Quantity"])
  return total


# 7. Find customer who spent the most money (Delivered orders only)
def top_customer_by_spend(orders_list, products_list, customers_list):
  spend_map = {}

  for order in orders_list:
    if order["Status"] == "Delivered":
      cust_id = order["Customer_ID"]
      price = get_product_price(order["Product_ID"], products_list)
      val = calculate_order_value(price, order["Quantity"])

      # Add to customer's total spend
      spend_map[cust_id] = spend_map.get(cust_id, 0) + val

  # Find the customer ID with maximum spend
  top_cust_id = max(spend_map, key=spend_map.get)
  return get_customer_name(top_cust_id, customers_list)


# 8. Bonus (*args): Apply multiple percentage discounts sequentially
def apply_offer(price, *discounts):
  final_price = price
  for discount in discounts:
    final_price -= final_price * (discount / 100)
  return final_price


# --- Testing and Verification ---

print("1 & 3. Price of P01 & Order Value for Qty 3:")
p_price = get_product_price("P01", products)
print(f"Price: ₹{p_price}, Total: ₹{calculate_order_value(p_price, 3)}\n")

print("2. Name of C04:")
print(f"Customer Name: {get_customer_name('C04', customers)}\n")

print("4. Classify Order Values:")
print(f"₹4000: {classify_order(4000)}")
print(f"₹1500: {classify_order(1500)}")
print(f"₹500: {classify_order(500)}\n")

print("5. Generate Invoice Sample (Order O101):")
print(generate_invoice(orders[0], products, customers))
print("\n" + "=" * 40 + "\n")

print("6. Total Revenue Check:")
rev = calculate_total_revenue(orders, products)
print(f"Total Revenue: ₹{rev}\n")

print("7. Top Customer by Spend (Delivered Orders):")
top_cust = top_customer_by_spend(orders, products, customers)
print(f"Top Spender: {top_cust}\n")

print("8. Apply Discounts (₹1000 with 10% then 5% off):")
discounted = apply_offer(1000, 10, 5)
print(f"Final Price: ₹{discounted}")
```

```python
# ---  calculations for the report ---

# 1. Total products & distinct categories
total_products = len(products)
categories_set = set()
for p in products:
  categories_set.add(p["Category"])
total_categories = len(categories_set)

# 2. Total customers & Premium/Regular split
total_customers = len(customers)
premium_count = 0
regular_count = 0
for c in customers:
  if c["Membership"] == "Premium":
    premium_count += 1
  else:
    regular_count += 1

# 3. Order status counts (Delivered, Pending, Cancelled)
delivered_count = 0
pending_count = 0
cancelled_count = 0

for o in orders:
  if o["Status"] == "Delivered":
    delivered_count += 1
  elif o["Status"] == "Pending":
    pending_count += 1
  elif o["Status"] == "Cancelled":
    cancelled_count += 1

# 4. Total Revenue & Average Order Value (Delivered orders only)
delivered_revenue = 0

for o in orders:
  if o["Status"] == "Delivered":
    price = get_product_price(o["Product_ID"], products)
    delivered_revenue += calculate_order_value(price, o["Quantity"])

avg_order_value = (
    delivered_revenue / delivered_count if delivered_count > 0 else 0
)

# 5. Top Customer by spend (Delivered orders only)
top_customer = top_customer_by_spend(orders, products, customers)

# 6. Out of stock products count
out_of_stock_count = 0
for p in products:
  if p["Stock"] == 0:
    out_of_stock_count += 1

# 7. Most popular product category among Delivered orders
category_counts = {}
for o in orders:
  if o["Status"] == "Delivered":
    # Find category for this order's product
    for p in products:
      if p["Product_ID"] == o["Product_ID"]:
        cat = p["Category"]
        category_counts[cat] = category_counts.get(cat, 0) + o["Quantity"]

most_popular_category = max(category_counts, key=category_counts.get)


# --- PRINT THE CONSOLIDATED REPORT ---

print("==================================================")
print("       URBANCART BUSINESS SUMMARY REPORT          ")
print("==================================================")
print(f"Product Catalog     : {total_products} Products across {total_categories} Categories")
print(f"Customer Base       : {total_customers} Total ({premium_count} Premium, {regular_count} Regular)")
print(f"Orders Breakup      : {len(orders)} Total ({delivered_count} Delivered, {pending_count} Pending, {cancelled_count} Cancelled)")
print("--------------------------------------------------")
print(f"Total Revenue       : ₹{delivered_revenue:,.2f} (Delivered Orders Only)")
print(f"Average Order Value : ₹{avg_order_value:,.2f}")
print(f"Top Customer        : {top_customer}")
print(f"Out of Stock Items  : {out_of_stock_count}")
print(f"Top Category        : {most_popular_category}")
print("==================================================")
```
