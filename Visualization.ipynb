```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the diamonds dataset
diamonds = sns.load_dataset('diamonds')
print(diamonds.head())
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=diamonds, x='carat', y='price', hue='cut', alpha=0.6)
plt.title("Diamond Carat vs Price by Cut Quality")
plt.xlabel("Carat")
plt.ylabel("Price ($)")
plt.show()
```

```python
# Sampling 500 rows so the pairplot renders instantly
diamonds_sample = diamonds.sample(n=500, random_state=42)

sns.pairplot(data=diamonds_sample, hue='cut')
plt.show()
```

```python
# Compute correlation matrix on numeric columns only
corr_diamonds = diamonds.corr(numeric_only=True)

# Generate the gradient heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_diamonds, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Diamond Features")
plt.show()
```
